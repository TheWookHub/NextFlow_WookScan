```python
import os, sys
sys.path.insert(0, os.path.abspath('./dolphyn'))
import dolphyn as D
print(D.__file__) 
```

```python
#protein_seq_file = ""
#prac_Coxsackievirus_from_bips.fa"
#PhageScan_pilot_proteinClusterReps.fa" # your protein collection input, our example

# Example: Convert Coxsackievirus fasta file from BIPS as Input
#D.convert_multi_fasta_seq_to_single_line("./data/prac_Coxsackievirus_from_bips.fa", "converted_input_from_coxsa.fa")

# Convert .csv file from BIPS as input to fasta format
D.input_csv_to_fasta("./data/sample_input_from_bips.csv", "sample_input_from_bips.fasta")

protein_seq_file = "./sample_input_from_bips.fasta"

prot_file = "dolphyn_peptides_sample.faa"
dna_file_noadap = "dolphyn_oligos_noadap_sample.fasta"
dna_file = "dolphyn_oligos_sample.fasta"

five_prime_adapter = "AGGAATTCCGCTGCGT"
three_prime_adapter = "ATGGTCACAGCTGTGC"
```

```python
# TAKES TIME
# find the epitopes

import time

start_time = time.time()
# set testrun = 0 to run all!
ge = D.findEpitopes(testrun = 5, protein_seq_file = protein_seq_file, epitile_size=15, epitope_probability_cutoff = 0.5)

end_time = time.time()

print(str(int(end_time-start_time)) + "s elapsed")
```

```python
# if you want to store the predicted epitopes:

D.saveGlobalEpitopes(ge, "epi_probas_sample.json")
ge = D.readGlobalEpitopes("epi_probas_sample.json")
```

```python
# Store Origin Proteins Information which contain Predict Epitope
D.storeOriginProteinInfoFromPredictEpitope("epi_probas_sample.json", "bips_input_sample.csv", file_format="csv", seq_field="origin_sequences")

D.storeOriginProteinInfoFromPredictEpitope("epi_probas_sample.json", "bips_input_sample.fa", file_format="fa", seq_field="origin_sequences")
```

```python
dolphynTiles, _ = D.peptideStitching(no_epis_per_tile = 3, linker_seq = "GGGGS" , global_epitopes = ge, return_unused_epis = False)
protein_hash = D.writeDolphynPeptidesToFAA(dolphynTiles, prot_file, hash_protein_names = True)
```

```python
revtrans_command = "pepsyn revtrans "+ prot_file + " " + dna_file_noadap
stream = os.popen(revtrans_command)
```

```python
# Add adapter sequences
dna_file_noadap = "./dolphyn_oligos_noadap_sample.fasta"

with open(dna_file_noadap, "r") as na:
    with open(dna_file, "w") as wa:
        for line in na:
            l = line.strip()
            if not line.startswith(">"):
                l = five_prime_adapter + l + three_prime_adapter
            wa.write(l+"\n")
```
